```python
import numpy as np
import pandas as pd

# 1. Load an e-commerce / online retail sales dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/anagrams.data"  # fallback or use standard online retail url
# Let's use a robust public online retail/sales sample dataset:
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/daily-total-female-births.csv"  # simple time-series, or let's create a realistic e-commerce dataframe:

# Creating a clean, realistic e-commerce sales dataframe for robust feature engineering
np.random.seed(42)
n_rows = 1000

data = {
    "CustomerID": np.random.randint(1000, 1050, size=n_rows),
    "OrderDate": pd.date_range(start="2025-01-01", periods=n_rows, freq="h"),
    "Category": np.random.choice(
        ["Electronics", "Clothing", "Home & Kitchen", "Books"], size=n_rows
    ),
    "Quantity": np.random.randint(1, 5, size=n_rows),
    "UnitPrice": np.random.uniform(10.0, 300.0, size=n_rows),
}

df = pd.DataFrame(data)
df["TotalSales"] = df["Quantity"] * df["UnitPrice"]
print("Initial Dataset Sample:")
display(df.head())

# 2. Feature Engineering & Date Extraction
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month_name()
df["DayOfWeek"] = df["OrderDate"].dt.day_name()
df["Hour"] = df["OrderDate"].dt.hour

print(
    "\nDataset with Extracted Date Features (Year, Month, DayOfWeek, Hour):"
)
display(df.head())

# 3. Pivot Tables & Trend Calculations (Sales trends by Category and Month)
sales_pivot = pd.pivot_table(
    df,
    values="TotalSales",
    index="Category",
    columns="Month",
    aggfunc="sum",
    fill_value=0,
)

print("\nPivot Table: Total Sales by Category and Month:")
display(sales_pivot)

# Customer level aggregation (Top customer segments)
customer_summary = (
    df.groupby("CustomerID")
    .agg(
        TotalSpent=("TotalSales", "sum"),
        TotalOrders=("Quantity", "count"),
        AverageOrderValue=("TotalSales", "mean"),
    )
    .reset_index()
)

top_customers = customer_summary.sort_values(
    by="TotalSpent", ascending=False
).head(5)
print("\nTop 5 Customer Segments by Total Spend:")
display(top_customers)

# 4. Actionable Business Recommendations Summary
print("\n" + "=" * 50)
print("ACTIONABLE BUSINESS RECOMMENDATIONS SUMMARY:")
print("=" * 50)
print(
    "1. Focus marketing budget on peak sales months identified in the pivot table."
)
print(
    "2. Implement loyalty rewards programs for top customer segments (top spenders)."
)
print(
    "3. Run targeted promotions during lower-performing hours/days to balance traffic."
)
```
