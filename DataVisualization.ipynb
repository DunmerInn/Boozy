```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set a clean custom style for the plots
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# 1. Load a real dataset (using the classic Titanic or Iris dataset for great visuals)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Clean missing values briefly so plots render cleanly
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

print("Dataset loaded and prepped successfully!\n")

# 2. Bar Chart (Categorical distribution)
plt.figure()
sns.countplot(
    data=df, x="Pclass", hue="Survived", palette="Set2"
)  # Custom color palette
plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()

# 3. Scatter Plot (Relationship between two numerical variables)
plt.figure()
sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    palette=["#ff7675", "#55efc4"],
    alpha=0.7,
)
plt.title("Scatter Plot: Age vs. Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

# 4. Correlation Heatmap (Numerical features correlation matrix)
plt.figure(figsize=(6, 4))
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=["number"])
correlation_matrix = numeric_df.corr()

sns.heatmap(
    correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.show()
```
